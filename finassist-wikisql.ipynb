```python
def make_sql(item, table):
    sql_data = item["sql"]

    select_col = table["header"][sql_data["sel"]]

    if sql_data["agg"] == 0:
        sql = "SELECT " + select_col
    else:
        agg = ["", "MAX", "MIN", "COUNT", "SUM", "AVG"]
        sql = "SELECT " + agg[sql_data["agg"]] + "(" + select_col + ")"

    if len(sql_data["conds"]) > 0:
        conditions = []

        for cond in sql_data["conds"]:
            col = table["header"][cond[0]]
            op = ["=", ">", "<"][cond[1]]
            value = str(cond[2]).replace("'", "''")

            conditions.append(
                col + " " + op + " '" + value + "'"
            )

        sql += " WHERE " + " AND ".join(conditions)

    return sql
```

```python
import pandas as pd
import json

test_path = "/kaggle/input/datasets/mathematican/wikisql-finassist/data/test.jsonl"
test_tables_path = "/kaggle/input/datasets/mathematican/wikisql-finassist/data/test.tables.jsonl"

test_data = pd.read_json(test_path, lines=True)
test_tables = pd.read_json(test_tables_path, lines=True)

print("Test examples:", test_data.shape)
print("Test tables:", test_tables.shape)

print(test_data.head())
print(test_tables.head())
```

```python
test_table_dict = {
    row["id"]: row.to_dict()
    for _, row in test_tables.iterrows()
}
```

```python
import torch
import pandas as pd
from tqdm.auto import tqdm

model.eval()

sample_data = test_data.head(100)

inputs_text = []
actual_sql = []

for _, row in sample_data.iterrows():
    item = row.to_dict()
    table = test_table_dict.get(item["table_id"])

    if table is None:
        continue

    schema = " | ".join(table["header"])

    input_text = (
        "translate English to SQL: "
        + "Question: " + item["question"]
        + " Schema: " + schema
    )

    inputs_text.append(input_text)
    actual_sql.append(make_sql(item, table))

predictions = []

batch_size = 16

for start in tqdm(range(0, len(inputs_text), batch_size)):
    batch = inputs_text[start:start + batch_size]

    inputs = tokenizer(
        batch,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=256
    )

    inputs = {
        key: value.to(model.device)
        for key, value in inputs.items()
    }

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=64
        )

    batch_predictions = tokenizer.batch_decode(
        outputs,
        skip_special_tokens=True
    )

    predictions.extend(batch_predictions)

correct = sum(
    pred.strip().lower() == actual.strip().lower()
    for pred, actual in zip(predictions, actual_sql)
)

print("Evaluated examples:", len(predictions))
print(f"Exact Match Accuracy: {correct / len(predictions) * 100:.2f}%"
      if predictions else "No examples evaluated.")
```

```python
def normalize_sql(sql):
    return " ".join(sql.lower().strip().split())
```

```python
actual = test_df["sql"].apply(normalize_sql)
predicted = test_df["predicted_sql"].apply(normalize_sql)

exact_match_accuracy = (actual == predicted).mean()

print("Exact Match Accuracy:", exact_match_accuracy)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score
```

```python
y_true = []
y_pred = []

for actual_sql, predicted_sql in zip(
    test_df["sql"],
    test_df["predicted_sql"]
):
    actual_tokens = normalize_sql(actual_sql).split()
    predicted_tokens = normalize_sql(predicted_sql).split()

    max_len = max(len(actual_tokens), len(predicted_tokens))

    actual_tokens += ["<PAD>"] * (max_len - len(actual_tokens))
    predicted_tokens += ["<PAD>"] * (max_len - len(predicted_tokens))

    y_true.extend(actual_tokens)
    y_pred.extend(predicted_tokens)
```

```python
precision = precision_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
```

```python
import json

def convert_to_sql(prediction, schema):
    prediction = prediction.strip()

    if not prediction.startswith("{"):
        prediction = "{" + prediction

    if not prediction.endswith("}"):
        prediction = prediction + "}"

    data = json.loads(prediction)
    columns = schema["header"]

    agg = ["", "MAX", "MIN", "COUNT", "SUM", "AVG"]
    operators = ["=", ">", "<"]

    col = columns[data["sel"]]

    if data["agg"] == 0:
        sql = "SELECT " + col
    else:
        sql = "SELECT " + agg[data["agg"]] + "(" + col + ")"

    conditions = []

    for cond in data["conds"]:
        col_name = columns[cond[0]]
        op = operators[cond[1]]
        value = str(cond[2]).replace("'", "''")

        conditions.append(col_name + " " + op + " '" + value + "'")

    if conditions:
        sql += " WHERE " + " AND ".join(conditions)

    return sql


sql_query = convert_to_sql(prediction, table)
print("Predicted SQL Query:")
print(sql_query)
```
